```
using System;
using System.Collections.Generic;
using System.Linq;

// Интерфейс предприятия
public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    List<IEmployee> GetEmployees();
    decimal CalculateTotalExpenses();
}

// Интерфейс поставщика
public interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType { get; }
}

// Интерфейс сотрудника
public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; }
}

public class Enterprise : IEnterprise
{
    private readonly List<ISupplier> _suppliers = new List<ISupplier>();
    private readonly List<IEmployee> _employees = new List<IEmployee>();

    public void AddSupplier(ISupplier supplier)
    {
        _suppliers.Add(supplier);
    }

    public void AddEmployee(IEmployee employee)
    {
        _employees.Add(employee);
    }

    public List<ISupplier> GetSuppliers()
    {
        return _suppliers;
    }

    public List<IEmployee> GetEmployees()
    {
        return _employees;
    }

    public decimal CalculateTotalExpenses()
    {
        decimal totalCost = _suppliers.Sum(s => s.GetCost());
        decimal totalSalary = _employees.Sum(e => e.GetSalary());
        return totalCost + totalSalary;
    }
}

// Реализация поставщика
public class Supplier : ISupplier
{
    public string Name { get; }
    public decimal Cost { get; }
    public string SupplierType { get; }

    public Supplier(string name, decimal cost, string supplierType)
    {
        Name = name;
        Cost = cost;
        SupplierType = supplierType;
    }

    public string GetName()
    {
        return Name;
    }
    public decimal GetCost()
    {
        return Cost;
    }
}

// Реализация сотрудника
public class Employee : IEmployee
{
    public string FullName { get; }
    public string Position { get; }
    public decimal Salary { get; }

    public Employee(string fullName, string position, decimal salary)
    {
        FullName = fullName;
        Position = position;
        Salary = salary;
    }

    public string GetFullName()
    {
        return FullName;
    }
    public decimal GetSalary()
    {
        return Salary;
    }
}


IEnterprise enterprise = new Enterprise();

// Добавляем поставщиков
enterprise.AddSupplier(new Supplier("Поставщик кухонной утвари", 40000, "Кухонная утварь"));
enterprise.AddSupplier(new Supplier("Поставщик мясной продукции", 150000, "Мясная продукция"));

// Добавляем сотрудников
enterprise.AddEmployee(new Employee("Виктор Баринов", "Шеф-повар", 90000));
enterprise.AddEmployee(new Employee("Дмитрий Нагиев", "Владелец", 0));

// Вывод списка поставщиков
Console.WriteLine("Список поставщиков:");
int i = 1;
foreach (var supplier in enterprise.GetSuppliers())
{
    Console.WriteLine($"{i++}. {supplier.GetName()} – Тип: {supplier.SupplierType} – Стоимость услуг: {supplier.GetCost()}");
}

// Вывод списка сотрудников
Console.WriteLine("\nСписок сотрудников:");
i = 1;
foreach (var employee in enterprise.GetEmployees())
{
    Console.WriteLine($"{i++}. {employee.GetFullName()} – Должность: {employee.Position} – Зарплата: {employee.GetSalary()}");
}

// Подсчёт общих расходов
decimal total = enterprise.CalculateTotalExpenses();
Console.WriteLine($"\nОбщие расходы: {total}");
```
