```
using System;
using System.Collections.Generic;

public interface IInstallable
{
    void Install();
}
public abstract class Component
{
    public float Cost {get; set;}

    public Component(float cost)
    {
        Cost = cost;
    }
    public virtual void GetInfo()
    {
        Console.WriteLine("Компонент");
    }
}

public class CPU : Component, IInstallable
{
    public string Name {get; set;}
    public CPU(float cost, string name) : base(cost)
    {
        Name = name;
    }
    public override void GetInfo()
    {
        Console.WriteLine($"Процессор: {Name}");
    }
    public void Install()
    {
        Console.WriteLine("Процессор установлен");
    }
}
public class RAM : Component, IInstallable
{
    public string Name {get; set;}
    public RAM(float cost, string name) : base(cost)
    {
        Name = name;
    }
    public override void GetInfo()
    {
        Console.WriteLine($"Оперативная память: {Name}");
    }
    public void Install()
    {
        Console.WriteLine("Оперативная память установлена");
    }
}
public class GPU : Component, IInstallable
{
    public string Name {get; set;}
    public GPU(float cost, string name) : base(cost)
    {
        Name = name;
    }
    public override void GetInfo()
    {
        Console.WriteLine($"Видеокарта: {Name}");
    }
    public void Install()
    {
        Console.WriteLine("Видеокарта установлена");
    }
}
public class HDD : Component, IInstallable
{
    public string Name {get; set;}
    public HDD(float cost, string name) : base(cost)
    {
        Name = name;
    }
    public override void GetInfo()
    {
        Console.WriteLine($"Жесткий диск: {Name}");
    }
    public void Install()
    {
        Console.WriteLine("Жесткий Диск установлен");
    }
}

public class Computer
{
    public List<Component> Components = new List<Component>();
    public void AddComponent<T> (T Component) where T: Component, IInstallable
    {
        Components.Add(Component);
    }
    public void RemoveComponent<T> (T Component) where T: Component, IInstallable
    {
        Components.Remove(Component);
    }
    public float GetTotalPrice()
    {
        float total = 0f;
        foreach(Component c in Components)
        {
            total += c.Cost;
        }
        return total;
    }
}
public static class ComputerFactory
{

    public static void AssemblePC(CPU cpu, RAM ram, GPU gpu, HDD hdd)
    {
        Console.WriteLine("Производится сборка компьютера...");
        Computer pc = new Computer();
        pc.AddComponent(cpu);
        cpu.Install();
        pc.AddComponent(ram);
        ram.Install();
        pc.AddComponent(gpu);
        gpu.Install();
        pc.AddComponent(hdd);
        hdd.Install();
        Console.WriteLine("...");
        Console.WriteLine("Компьютер собран!");
        Console.WriteLine("Характеристики:");
        foreach (Component c in pc.Components)
        {
            c.GetInfo();
        }
        Console.WriteLine("...");
        Console.WriteLine($"Стоимость компьютера: {pc.GetTotalPrice()} руб.");
    }
}

CPU cpu = new CPU(39000.00f, "Ryzen 7 7800X3D");
RAM ram = new RAM(12000.00f, "2x ADATA 16 GB DDR5 6000Mhz");
GPU gpu = new GPU(250000.00f, "MSI GeForce RTX 5090");
HDD hdd = new HDD(15000.00f, "SeaGate 10TB SATA");
ComputerFactory.AssemblePC(cpu,ram,gpu,hdd);
```
