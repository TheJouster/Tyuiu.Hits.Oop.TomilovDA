```
using System;
using System.Collections.Generic;

// БАЗОВЫЙ КЛАСС
public class Task
{
    public int TaskId { get; set; }
    public string TaskName { get; set; }
    public string Priority { get; set; }
    public bool IsCompleted { get; set; }
    public string AssignedTo { get; set; }

    // Новые атрибуты
    public string Category { get; set; }
    public int EstimatedHours { get; set; }
    public double Progress { get; set; }
    public List<string> Tags { get; set; }

    public string Description { get; set; }
    public DateTime CreatedAt { get; set; }
    public List<string> Comments { get; set; }

    public Task(int taskId, string taskName, string priority, string assignedTo)
    {
        TaskId = taskId;
        TaskName = taskName;
        Priority = priority;
        AssignedTo = assignedTo;
        IsCompleted = false;
        CreatedAt = DateTime.Now;
        Comments = new List<string>();
        Tags = new List<string>();
        Progress = 0;
    }

    // ПОЛИМОРФИЗМ (переопределяемый метод)
    public virtual void MarkAsComplete()
    {
        IsCompleted = true;
        Progress = 100;
        Console.WriteLine($"Задача '{TaskName}' выполнена.");
    }

    public virtual void GetTaskDetails()
    {
        Console.WriteLine($"ID: {TaskId}, Название: {TaskName}, Приоритет: {Priority}, Исполнитель: {AssignedTo}, Статус: {(IsCompleted ? "выполнена" : "в процессе")}");
        Console.WriteLine($"Прогресс: {Progress}% | Категория: {Category ?? "Не указана"} | Оценка: {EstimatedHours} часов");
    }

    public virtual void ReassignTo(string newAssign)
    {
        AssignedTo = newAssign;
        Console.WriteLine($"Задача '{TaskName}' переназначена исполнителю: {newAssign}");
    }

    //Перегрузка метода
    public void ReassignTo(string newAssign, string reason)
    {
        ReassignTo(newAssign);
        Console.WriteLine($"Причина переназначения: {reason}");
    }

    public void AddComment(string comment)
    {
        Comments.Add(comment);
        Console.WriteLine($"Комментарий добавлен: \"{comment}\"");
    }

    public void ShowComments()
    {
        Console.WriteLine("Комментарии:");
        foreach (var c in Comments)
            Console.WriteLine($" - {c}");
    }

    // Новый метод
    public void AddTag(string tag)
    {
        Tags.Add(tag);
        Console.WriteLine($"Добавлен тег: {tag}");
    }
}

// ПРОИЗВОДНЫЙ КЛАСС 1
public class DelegateTask : Task
{
    public DateTime DueDate { get; set; }
    public int ReminderDays { get; set; }
    public bool IsUrgent { get; set; }  // новый атрибут
    public string Supervisor { get; set; }

    public DelegateTask(int taskId, string taskName, string priority, string assignedTo, DateTime dueDate)
        : base(taskId, taskName, priority, assignedTo)
    {
        DueDate = dueDate;
        ReminderDays = 1;
    }

    // Переопределение
    public override void MarkAsComplete()
    {
        base.MarkAsComplete();
        Console.WriteLine($"Дата завершения задачи: {DateTime.Now.ToShortDateString()} (срок был до {DueDate.ToShortDateString()})");
    }

    // Перегрузка метода
    public void SetReminder(int daysBefore)
    {
        ReminderDays = daysBefore;
        Console.WriteLine($"Напоминание установлено за {ReminderDays} дней до срока.");
    }

    public void SetReminder(DateTime customDate)
    {
        Console.WriteLine($"Напоминание установлено на {customDate.ToShortDateString()}.");
    }
}

//ПРОИЗВОДНЫЙ КЛАСС 2
public class TeamTask : Task
{
    public string TeamName { get; set; }
    public List<string> TeamMembers { get; set; }
    public string ProjectName { get; set; }  // новый атрибут

    public TeamTask(int taskId, string taskName, string priority, string assignedTo, string teamName)
        : base(taskId, taskName, priority, assignedTo)
    {
        TeamName = teamName;
        TeamMembers = new List<string>();
    }

    // Переопределение метода
    public override void ReassignTo(string newAssignee)
    {
        base.ReassignTo(newAssignee);
        Console.WriteLine($"Задача теперь назначена члену команды '{TeamName}': {newAssignee}");
    }

    // Новый метод
    public void AddMember(string member)
    {
        TeamMembers.Add(member);
        Console.WriteLine($"Добавлен новый участник в команду '{TeamName}': {member}");
    }

    public void ShowTeam()
    {
        Console.WriteLine($"Команда '{TeamName}': {string.Join(", ", TeamMembers)}");
    }
}

//ПРОИЗВОДНЫЙ КЛАСС 3
public interface IResearchable
{
    void AnalyzeData();
    void GenerateReport();
}

public class ResearchTask : Task, IResearchable
{
    public string DataSource { get; set; }
    public int DataVolumeMB { get; set; } // новый атрибут

    public ResearchTask(int taskId, string taskName, string priority, string assignedTo, string dataSource)
        : base(taskId, taskName, priority, assignedTo)
    {
        DataSource = dataSource;
    }

    public override void GetTaskDetails()
    {
        base.GetTaskDetails();
        Console.WriteLine($"Источник данных: {DataSource}, Объём: {DataVolumeMB} МБ");
    }

    public void AnalyzeData()
    {
        Console.WriteLine($"Анализ данных из источника: {DataSource}");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Создан отчёт по результатам анализа задачи '{TaskName}'.");
    }
}

// GENERIC-КЛАСС
public class TaskManager<T> where T : Task
{
    private List<T> tasks = new List<T>();

    public void AddTask(T task)
    {
        tasks.Add(task);
        Console.WriteLine($"Добавлена задача: {task.TaskName}");
    }

    public void ShowAllTasks()
    {
        Console.WriteLine("\n=== Список задач ===");
        foreach (var t in tasks)
            t.GetTaskDetails();
    }

    public void CompleteAll()
    {
        foreach (var t in tasks)
            t.MarkAsComplete();
    }
}


var delegateTask = new DelegateTask(1, "Подготовить отчёт", "Высокий", "Дмитрий", DateTime.Now.AddDays(2));
delegateTask.SetReminder(2);
delegateTask.SetReminder(DateTime.Now.AddDays(1));
delegateTask.AddTag("отчёт");
delegateTask.MarkAsComplete();

Console.WriteLine();

var teamTask = new TeamTask(2, "Разработка модуля", "Средний", "Кирилл", "DevTeam");
teamTask.AddMember("Сергей");
teamTask.AddMember("Анна");
teamTask.ReassignTo("Анна");
teamTask.GetTaskDetails();

Console.WriteLine();

var researchTask = new ResearchTask(3, "Анализ рынка", "Низкий", "Максим", "Открытые источники");
researchTask.AnalyzeData();
researchTask.GenerateReport();

Console.WriteLine();

var manager = new TaskManager<Task>();
manager.AddTask(delegateTask);
manager.AddTask(teamTask);
manager.AddTask(researchTask);
manager.ShowAllTasks();
```
