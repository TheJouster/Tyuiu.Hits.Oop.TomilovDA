```
using System;

public class Plant
{
    public string Name {get; set;}
    public string Type {get; set;}
    public float Height {get; set;}

    public Plant(string name, float height)
    {
        Name = name;
        Type = "Растение";
        Height = height;
    }
    public virtual void Bloom()
    {
        Console.WriteLine($"{Type} {Name} цветёт");
    }
    public void Grow(float height)
    {
        Height += height;
        Console.WriteLine($"{Type} выросло на {height}. Высота: {Height}");
    }
    
}

public class Flower : Plant
{
    public string Color {get; set;}
    public int PetalCount {get; set;}

    public Flower(string name, float height, string color, int petalCount) : base(name, height)
    {
        Type = "Цветок";
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        base.Bloom();
        Console.WriteLine($"Цветок имеет {PetalCount} лепестков и {Color} цвет.");
    }

    public void Fade()
    {
        Console.WriteLine("Цветок увядает:(");
    }
}

public class Tree : Plant
{
    public int Age {get; set;}
    public string LeafType {get; set;}

    public Tree(string name, float height, int age, string leafType) : base(name, height)
    {
        Type = "Дерево";
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        base.Bloom();
        Console.WriteLine($"Возраст дерева {Age} лет и оно имеет {LeafType} листья.");
    }

    public void GrowUp(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево росло {years} лет. Теперь дереву {Age} лет.");
        base.Grow(years*10);
    }
}

Flower myFlower = new Flower("Роза", 2, "красный", 18);
myFlower.Bloom();
myFlower.Grow(2);
myFlower.Fade();

Console.WriteLine();

Tree myTree = new Tree("Сосна", 18, 5, "хвойные");
myTree.Bloom();
myTree.Grow(20);
myTree.GrowUp(5);
```
