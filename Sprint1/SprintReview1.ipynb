```
using System;
using System.Collections.Generic;

public class Product
{
    private string Name {get; set;}
    private decimal Price {get; set;}
    private int Quantity {get; set;}
    protected string Category {get; set;}
    internal string Description {get; set;}

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }
    public decimal GetTotalPrice()
    {
        return Price;
    }

    public void UpdateQuantity(int amount)
    {
        if (amount > 0)
        {
            Console.WriteLine($"Товар {Name} пополнился на {amount} ед.");
        }
        Quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"Товар {Name} имеетеся в наличии {Quantity} штук. Цена товара: {Price} рублей. Описание: {Description}";
    }
}

public class Seller
{
    private string Name {get; set;}
    private string EmployeeId {get; set;}
    protected decimal Salary {get; set;}
    internal string ContactInfo {get; set;}
    public Store MyStore {get; set;}

    public Seller(string name, string id, decimal salary, string info)
    {
        Name = name;
        EmployeeId = id;
        Salary = salary;
        ContactInfo = info;
    }

    public void AddProduct(Product product)
    {
        MyStore.Products.Add(product);
        //Console.WriteLine(product.GetProductInfo());
    }

    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);
        Console.WriteLine($"Продавец продал {quantity} штук товара");
        Console.WriteLine(product.GetProductInfo());
    }

    public string GetSellerInfo()
    {
        return $"Сотрудник #{EmployeeId} {Name}. Зарплата: {Salary} рублей. Номер телефона: {ContactInfo}";
    }

}

public class Store
{
    private string StoreName {get; set;}
    private string Location {get; set;}
    public string StoreHours {get; set;}
    public List<Product> Products {get; set;}
    public List<Seller> Sellers {get; set;}

    public Store(string name, string location, string storeHours)
    {
        StoreName = name;
        Location = location;
        StoreHours = storeHours;
        Products = new List<Product>();
        Sellers = new List<Seller>();
    }
    public void AddSeller(Seller seller)
    {
        seller.MyStore = this;
        Sellers.Add(seller);
        
        
    }
    public void ListProducts()
    {
        Console.WriteLine("Продукты:");
        foreach (var p in Products)
        {
            Console.WriteLine(p.GetProductInfo());
        }
    }
    public void ListSellers()
    {
        Console.WriteLine("Продавцы:");
        foreach (var p in Sellers)
        {
            Console.WriteLine(p.GetSellerInfo());
        }
    }
    public string GetStoreInfo()
    {
        return $"Магазин {StoreName} находящийся на {Location}, открыт с {StoreHours}";
    }
}

Store myStore = new Store("Ашан", "ул. Горького 42", "с 8 до 20");
Seller mySeller = new Seller("Гена", "007", 34255, "89524924574");
Product apple = new Product("Яблоко", 25, 321, "Фрукты", "Красные сочные яблоки");

Console.WriteLine(myStore.GetStoreInfo());
myStore.AddSeller(mySeller);
myStore.ListSellers();
mySeller.AddProduct(apple);
myStore.ListProducts();
mySeller.SellProduct(apple, 20);

```
