```
using System;
using System.Collections.Generic;

public interface IReportable
{
    string GenerateReport();
}
public interface ITeamLeader
{
    List<String> GetTeamMembers();
}

public class Employee 
{
    public string Name {get;set;}
    public string Position {get; set;}
    public float Salary {get; set;}

    public Employee(string name,string position,float salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }
    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник {Name} на позиции {Position} с зарплатой {Salary} руб.");
    }
}

public class Manager : Employee
{
    public string Department {get;set;}

    public Manager(string name,string position,float salary, string department):base(name,position,salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        Console.WriteLine($"Сотрудник {Name} на позиции {Position} с зарплатой {Salary} руб. в департаменте {Department}");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName {get; set;}
    public override void PrintInfo()
    {
        Console.WriteLine($"Сотрудник {Name} на позиции {Position} с зарплатой {Salary} руб. в департаменте {Department} назначен на проект {ProjectName}");
    }
    public string GenerateReport()
    {
        return $"{ProjectName} проект идет по плану. Отчет окончен";
    }
    public ProjectManager(string name,string position,float salary, string department, string project):base(name,position,salary,department)
    {
        ProjectName = project;
    }
    public List<String> GetTeamMembers()
    {
        List<String> Members = new List<String>();
        Members.Add("Программист");
        Members.Add("Разработчик");
        Members.Add("Тимлид");
        return Members;
    }
}

Employee emp1 = new Employee("Деметрий","Программист",696969.69f);
emp1.PrintInfo();
Manager emp2 = new Manager("Максим","Разработчик",12344.65f,"Максимов");
emp2.PrintInfo();
ProjectManager emp3 = new ProjectManager("Кирилл","Тимлид",1.65f,"Кириллов","SkillCards");
emp3.PrintInfo();
emp3.GenerateReport();
List<String> members = emp3.GetTeamMembers();
Console.WriteLine("Участники команды:");
foreach(string c in members)
{
    Console.WriteLine(c);
}
```
